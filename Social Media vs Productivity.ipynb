# SOCIAL MEDIA HABITS AND PRODUCTIVITY

## Project Introduction

The Students' Social Media Habits vs Productivity project aims to analyze how students' daily social media usage and digital habits influence their academic productivity, concentration, and overall lifestyle.
The project seeks to identify patterns and relationships between excessive social media usage, digital distractions, and students' academic performance.
The analysis will use data analytics and visualization techniques to uncover meaningful insights and understand which lifestyle habits are associated with higher or lower productivity.

## Project Objectives

The main objectives of this project are:
- Analyze the relationship between social media usage and student productivity.
- Identify whether excessive screen time is associated with reduced focus.
- Examine how study hours and sleep duration relate to productivity.
- Investigate the impact of notifications on students' concentration.
- Understand how nighttime social media usage affects productivity.
- Explore the relationship between addiction levels and distraction scores.
- Identify lifestyle habits associated with better productivity.
- Provide data-driven recommendations for improving students' study habits.

## Buisness problem

Students often struggle to balance their academic responsibilities with their social media activities. Spending excessive time on social media may reduce study time, interrupt concentration through frequent notifications, and affect sleep quality.

However, not all students who use social media extensively experience low productivity. Factors such as study habits, sleep duration, breaks, and self-control may also influence productivity.

Therefore, this project investigates:
- How does social media usage affect students' productivity?
- Is there a relationship between daily screen time and study hours?
- Does excessive social media usage lead to higher distraction scores?
- How does nighttime social media usage affect sleep hours?
- Do frequent notifications negatively influence focus?
- Is addiction level associated with lower productivity?
- Which factors contribute most significantly to students' productivity?

## Dataset Description

The dataset consists of 13 variables describing students' demographics, digital habits, lifestyle patterns, and academic productivity
It contains 10000 records.

|Field name|Description|
|:---|:---|
|age|Age of the student|
|daily_screen_time|Total daily screen time in hours|
|social_media_hours|Daily hours spent on social media|
|study_hours|Number of hours spent studying daily|
|sleep_hours|Average daily sleeping hours|
|notifications_per_day|Number of notifications received daily|
|coffee_per_day|Number of coffees consumed per day|
|breaks_per_day|Number of breaks taken daily|
|focus_score|Student's ability to maintain concentration|
|night_usage|Indicates whether the student uses social media at night|
|distraction_score|Level of distraction experienced by the student|
|productivity_score|Overall productivity level of the student|
|addiction_level|Level of social media dependency|

## Data Analysis Methodology

The project will follow these major stages:

1. **Data Collection**

Import the dataset from a CSV or Excel file using Python.

2. **Data Cleaning**

- Identify missing values.

- Remove duplicate records.

- Check incorrect or inconsistent values.

- Identify and investigate outliers.

- Correct data types and categorical values.

3. **Exploratory Data Analysis (EDA)**

- Calculate mean, median, mode, standard deviation, minimum, and maximum.

- Examine the distributions of important variables.

- Identify patterns and trends.

4. **Data Visualization**

- Create charts to identify trends, comparisons, distributions, and relationships.

5. **Correlation Analysis**

- Examine relationships between social media habits, lifestyle factors, distraction, focus, and productivity.

6. **Statistical Analysis**

- Apply appropriate statistical tests and regression analysis to investigate the research hypotheses.

7. **Interpretation and Conclusion**

- Summarize findings and provide recommendations for improving students' productivity.

## Project Limitations
- Dataset Dependency: Results depend on the accuracy and completeness of the available data.
- Limited Variables: Factors such as stress, motivation, and academic performance are not included.
- Self-Reported Data: Students' responses may not always be completely accurate.
- Correlation vs. Causation: The analysis cannot prove that social media directly causes changes in productivity.
- Limited Generalization: Findings may not represent all students or educational backgrounds.
- Subjective Scores: Focus, distraction, and productivity scores may depend on how they are measured.
- External Factors: Personal circumstances and study environments may also influence productivity.
