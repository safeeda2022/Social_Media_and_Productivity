## Project Introduction

The Students' Social Media Habits vs Productivity project aims to analyze how students' daily social media usage and digital habits influence their academic productivity, concentration, and overall lifestyle.
The project seeks to identify patterns and relationships between excessive social media usage, digital distractions, and students' academic performance.
The analysis will use data analytics and visualization techniques to uncover meaningful insights and understand which lifestyle habits are associated with higher or lower productivity.

## Project Objectives

The main objectives of this project are:
- Analyze the relationship between social media usage and student productivity.
- Identify whether excessive screen time is associated with reduced focus.
- Examine how study hours and sleep duration relate to productivity.
- Investigate the impact of notifications on students' concentration.
- Understand how nighttime social media usage affects productivity.
- Explore the relationship between addiction levels and distraction scores.
- Identify lifestyle habits associated with better productivity.
- Provide data-driven recommendations for improving students' study habits.

## Buisness problem

Students often struggle to balance their academic responsibilities with their social media activities. Spending excessive time on social media may reduce study time, interrupt concentration through frequent notifications, and affect sleep quality.

However, not all students who use social media extensively experience low productivity. Factors such as study habits, sleep duration, breaks, and self-control may also influence productivity.

Therefore, this project investigates:
- How does social media usage affect students' productivity?
- Is there a relationship between daily screen time and study hours?
- Does excessive social media usage lead to higher distraction scores?
- How does nighttime social media usage affect sleep hours?
- Do frequent notifications negatively influence focus?
- Is addiction level associated with lower productivity?
- Which factors contribute most significantly to students' productivity?

### Outliers
From the statistical data the field "notifications_per_day"is observed to have unrealistic values. Also the feilds "distraction_score" and "productivity_score" consists of outliers and does not follow a score range.

```python
col1=["notifications_per_day","distraction_score","productivity_score"]
```

```python
Q1=df[col1].quantile(0.25)
```

```python
Q3=df[col1].quantile(0.75)
```

```python
IQR=Q3-Q1
```

```python
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
```

```python
outlier_mask=(df[col1]<lower)|(df[col1]>upper)
```

```python
outlier_rows=df[outlier_mask.any(axis=1)]
outlier_rows
```

```python
print(outlier_mask.sum())
```

The outlier values are treated by the lower and upper bound obtained through IQR method.

```python
df[col1]=df[col1].clip(lower=lower,upper=upper,axis=1)
```

```python
df["notifications_per_day"]=df["notifications_per_day"].clip(lower=0)
```

```python
df["productivity_score"]=df["productivity_score"].clip(lower=0,upper=100)
```

```python
df["distraction_score"]=df["distraction_score"].clip(lower=0,upper=10)
```

The range of the fields in col1 are changed appropriately.
- productivity_score - 0 to 100
- distraction_score - 0 to 10
- lower bound of notifications_per_day to 0.

```python
df.describe()
```

### Null values

```python
col_mean=["age","daily_screen_time","social_media_hours","study_hours","sleep_hours"]
```

```python
col_median=["notifications_per_day","focus_score","coffee_per_day","breaks_per_day","distraction_score","productivity_score"]
```

```python
df[col_mean]=df[col_mean].fillna(df[col_mean].mean())
df[col_median]=df[col_median].fillna(df[col_median].median())
```

```python
df["night_usage"]=df["night_usage"].fillna(df["night_usage"].mode()[0])
```

```python
df["addiction_level"]=df["addiction_level"].fillna(df["addiction_level"].mode()[0])
```

The null values are imputed with statistical measures.
- The field with almost normal distributions are imputed with mean where as those with outliers are imputed with median.
- The two categorical fields are imputed with mode.

### Data Type

```python
df["age"]=df["age"].astype(int)
```

```python
df["notifications_per_day"]=df["notifications_per_day"].round().astype(int)
```

```python
df["coffee_per_day"]=df["coffee_per_day"].round().astype(int)
```

```python
df["breaks_per_day"]=df["breaks_per_day"].round().astype(int)
```

```python
df["night_usage"]=df["night_usage"].astype(int)
```

```python
df=df.round(2)
```

The data type of the fields are changed appropriately.
- Age,notifications_per_day,coffee_per_day,breaks_per_day,night_usage are converted to int.
- All the score fields are rounded to 2 decimals.
